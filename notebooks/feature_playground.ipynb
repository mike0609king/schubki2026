```python
def tsne_visualize(X, y, title):
    X_2d = TSNE(n_components=2 ,perplexity=30, random_state=42).fit_transform(X)
    classes = np.unique(y)
    for cls in classes:
        idx = y == cls
        plt.scatter(
            X_2d[idx, 0],
            X_2d[idx, 1],
            s=7,
        )
    plt.title(title)
feature_dir = Path("/content/drive/MyDrive/eeg_features/")
X, Y = load_features(feature_dir / "emd_biot" / "features.h5")
print(X.shape)
print(Y.shape)

tsne_visualize(X, Y, "")
```
